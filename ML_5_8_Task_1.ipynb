```python
# Import libraries
import pandas as pd
from sklearn.linear_model import LinearRegression

# Sample dataset
data = {
    "Distance_km": [2, 5, 8, 10, 12, 15, 18, 20],
    "Duration_min": [8, 15, 20, 25, 30, 35, 40, 45],
    "Fare": [8, 15, 22, 28, 33, 40, 46, 52]
}

df = pd.DataFrame(data)

print("Taxi Dataset")
print(df)

# Features
X = df[["Distance_km", "Duration_min"]]

# Target
y = df["Fare"]

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict fare
distance = 9
duration = 22

predicted_fare = model.predict([[distance, duration]])

print("\nPredicted Taxi Fare")
print(f"Distance: {distance} km")
print(f"Duration: {duration} minutes")
print(f"Estimated Fare: ${predicted_fare[0]:.2f}")
```

```python
import pandas as pd
import numpy as np

# Create the dataset from the markdown table
data = {
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class': ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
}
df = pd.DataFrame(data)
display(df)
```

```python
def euclidean_distance(point1, point2):
    """Calculates the Euclidean distance between two points."""
    # Ensure points are numpy arrays for element-wise operations
    point1 = np.array(point1)
    point2 = np.array(point2)
    return np.sqrt(np.sum((point1 - point2)**2))
```

```python
# Define a new entry
new_entry = [45, 40] # Brightness, Saturation

# Calculate distances from the new entry to all existing data points
df['Distance'] = df.apply(lambda row: euclidean_distance(new_entry, [row['Brightness'], row['Saturation']]), axis=1)

display(df)
```

```python
# Random Forest - Majority Voting

from collections import Counter

# Predictions made by different decision trees
tree_predictions = ["Apple", "Apple", "Banana"]

print("Predictions from individual trees:")
for i, prediction in enumerate(tree_predictions, 1):
    print("Tree", i, "->", prediction)

# Majority voting
votes = Counter(tree_predictions)

final_class = votes.most_common(1)[0][0]

print("\nVotes:")
for fruit, count in votes.items():
    print(fruit, "->", count)

print("\nFinal Class:", final_class)
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Simple fruit dataset
data = {
    "Color": [1, 1, 0, 0, 1, 0, 1, 0],
    "Size":  [1, 0, 1, 0, 1, 0, 0, 1],
    "Weight":[1, 1, 0, 0, 1, 0, 1, 0],
    "Fruit": ["Apple", "Apple", "Banana", "Banana",
              "Apple", "Banana", "Apple", "Banana"]
}

df = pd.DataFrame(data)

# Input features
X = df[["Color", "Size", "Weight"]]

# Target
y = df["Fruit"]

# Create Random Forest
model = RandomForestClassifier(
    n_estimators=10,
    random_state=42
)

# Train
model.fit(X, y)

# New fruit sample
new_fruit = [[1, 1, 1]]

# Prediction
prediction = model.predict(new_fruit)

print("Predicted Fruit:", prediction[0])
```

```python
import pandas as pd
from sklearn.svm import SVC

# Sample dataset
data = {
    "Word_Frequency":[12,3,15,1,10,2,20,4],
    "Message_Length":[300,50,450,40,350,60,500,80],
    "Spam":[1,0,1,0,1,0,1,0]
}

df = pd.DataFrame(data)

print(df)
```

```python
X = df[["Word_Frequency","Message_Length"]]

# Target
y = df["Spam"]

# Train SVM
model = SVC(kernel='linear')

model.fit(X,y)

# Test email
new_email = [[8,250]]

prediction = model.predict(new_email)

if prediction[0] == 1:
    print("\nPrediction: Spam Email")
else:
    print("\nPrediction: Not Spam")
```

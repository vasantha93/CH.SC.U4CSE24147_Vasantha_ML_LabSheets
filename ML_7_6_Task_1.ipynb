```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
```

```python
data = {
    'Age': [22, 25, 28, 32, 35, 40, 45, 50, 29, 38],
    'Income': ['Low', 'Low', 'Medium', 'Medium', 'High', 'High', 'Medium', 'High', 'Low', 'Medium'],
    'Buy_House': ['No', 'No', 'No', 'Yes', 'Yes', 'Yes', 'Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

# Convert Income into numbers
encoder = LabelEncoder()
df['Income'] = encoder.fit_transform(df['Income'])
```

```python
X = df[['Age', 'Income']]
y = df['Buy_House']

# Train Decision Tree
model = DecisionTreeClassifier(criterion='entropy', random_state=42)
model.fit(X, y)
```

```python
new_customer = [[30, encoder.transform(['Medium'])[0]]]
prediction = model.predict(new_customer)

print("Prediction:", prediction[0])
```

```python
plt.figure(figsize=(10,6))
plot_tree(model,
          feature_names=['Age', 'Income'],
          class_names=model.classes_,
          filled=True)

plt.title("Decision Tree - House Purchase")
plt.show()
```
